# Feature Engineering

```python
df["date"] = pd.to_datetime(df["date"])

df["sale_year"] = df.date.dt.year

df["sale_month"] = df.date.dt.month

df.drop("date", axis=1, inplace=True)
```

```python
df.drop("id", axis=1, inplace=True)
```

```python
df["house_age"] = df["sale_year"] - df["yr_built"]
```

```python
import numpy as np

df["renovated"] = (df.yr_renovated > 0).astype(int)

df["years_since_renovation"] = np.where(
    df["yr_renovated"] == 0,
    df["house_age"],
    df["sale_year"] - df["yr_renovated"]
)

df.drop("yr_renovated", axis=1, inplace=True)
```

```python
df["log_sqft_living"] = np.log1p(
    df["sqft_living"]
)
```

```python
df["grade_living"] = (
    df["grade"] *
    df["sqft_living"]
)
```

# Models

```python
from sklearn.model_selection import train_test_split

X = df.drop("price", axis=1)
y = np.log1p(df["price"])

X["zipcode"] = X["zipcode"].astype(str)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
numeric_features = X_train.select_dtypes(
    include=[
        "int64",
        "float64"
    ]
).columns


categorical_features = X_train.select_dtypes(
    include="object"
).columns
```

```python
from sklearn.pipeline import Pipeline

from sklearn.compose import ColumnTransformer

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

numeric_pipeline = Pipeline(
[
    (
        "scaler",
        StandardScaler()
    )
]
)


categorical_pipeline = Pipeline(
[
    (
        "encoder",
        OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    )
]
)

preprocessor = ColumnTransformer(
[
    (
        "num",
        numeric_pipeline,
        numeric_features
    ),

    (
        "cat",
        categorical_pipeline,
        categorical_features
    )
]
)
```

```python
from sklearn.linear_model import (
    LinearRegression,
    Ridge
)


from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor,
    HistGradientBoostingRegressor
)


from sklearn.tree import DecisionTreeRegressor


from xgboost import XGBRegressor

from lightgbm import LGBMRegressor

models = {


"LinearRegression":
LinearRegression(),


"Ridge":
Ridge(),


"DecisionTree":
DecisionTreeRegressor(
    random_state=42
),


"RandomForest":
RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
),


"GradientBoosting":
GradientBoostingRegressor(),


"HistGradientBoosting":
HistGradientBoostingRegressor(),


"XGBoost":
XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    random_state=42
),


"LightGBM":
LGBMRegressor(
    n_estimators=500,
    learning_rate=0.05,
    random_state=42
)

}
```

```python
from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_validate
)

cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


results=[]
```

```python
for name,model in models.items():


    pipe = Pipeline(
    [
        (
            "preprocessor",
            preprocessor
        ),

        (
            "model",
            model
        )
    ]
    )


    scores = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=cv,
        scoring=[
            "r2",
            "neg_root_mean_squared_error",
            "neg_mean_absolute_error"
        ],
        n_jobs=-1
    )


    results.append(
    {
        "Model":name,

        "R2":
        scores["test_r2"].mean(),

        "RMSE":
        -scores["test_neg_root_mean_squared_error"].mean(),

        "MAE":
        -scores["test_neg_mean_absolute_error"].mean()
    }
    )
```

```python
from catboost import CatBoostRegressor


cat_features = [
    "zipcode"
]



for col in cat_features:
    X_train[col] = X_train[col].astype(str)
    X_test[col] = X_test[col].astype(str)


cat_features_index = [
    X_train.columns.tolist().index(col)
    for col in cat_features
]


cat_model = CatBoostRegressor(

    iterations=1000,

    learning_rate=0.03,

    depth=8,

    loss_function="RMSE",

    random_seed=42,

    verbose=0

)



cat_scores = cross_validate(

    cat_model,

    X_train,

    y_train,

    cv=cv,

    scoring=[
        "r2",
        "neg_root_mean_squared_error",
        "neg_mean_absolute_error"
    ],

    params={
        "cat_features":cat_features_index
    },

    n_jobs=-1

)


results.append({

    "Model":"CatBoost",

    "R2":
    cat_scores["test_r2"].mean(),

    "RMSE":
    -cat_scores["test_neg_root_mean_squared_error"].mean(),

    "MAE":
    -cat_scores["test_neg_mean_absolute_error"].mean()

})

```

```python
cv_results = pd.DataFrame(results)


cv_results = cv_results.sort_values(
    by="R2",
    ascending=False
)


cv_results.reset_index(drop=True)
```

```python
def objective(trial):

    params = {

        "iterations":
        trial.suggest_int(
            "iterations",
            500,
            2000
        ),

        "depth":
        trial.suggest_int(
            "depth",
            5,
            10
        ),

        "learning_rate":
        trial.suggest_float(
            "learning_rate",
            0.01,
            0.1,
            log=True
        ),

        "l2_leaf_reg":
        trial.suggest_float(
            "l2_leaf_reg",
            1,
            10
        ),

        "random_strength":
        trial.suggest_float(
            "random_strength",
            0,
            2
        )
    }


    model = CatBoostRegressor(

        **params,

        loss_function="RMSE",

        random_seed=42,

        verbose=0
    )


    scores = cross_validate(

        model,

        X_train,

        y_train,

        cv=cv,

        scoring="r2",

        params={
            "cat_features":cat_features_index
        },

        n_jobs=-1
    )


    return scores["test_score"].mean()
```

```python
import optuna

study = optuna.create_study(
    direction="maximize"
)


study.optimize(
    objective,
    n_trials=50
)


study.best_params
```

{'iterations': 1583,

 'l2_leaf_reg': 2.8627508663139265,
 'random_strength': 0.6002641168992378}

```python
final_model = CatBoostRegressor(

    **study.best_params,

    loss_function="RMSE",

    random_seed=42,

    verbose=0

)


final_model.fit(

    X_train,

    y_train,

    cat_features=cat_features_index

)
```

```python
import os

os.makedirs(
    "models",
    exist_ok=True
)

final_model.save_model(
    "models/catboost_house_price_model.cbm"
)
```

```python
pred_log = final_model.predict(X_test)

pred_price = np.expm1(pred_log)

real_price = np.expm1(y_test)


from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error,
    mean_absolute_percentage_error
)

test_r2 = r2_score(
    real_price,
    pred_price
)


test_rmse = np.sqrt(
    mean_squared_error(
        real_price,
        pred_price
    )
)


test_mae = mean_absolute_error(
    real_price,
    pred_price
)


test_mape = mean_absolute_percentage_error(
    real_price,
    pred_price
)

final_report = pd.DataFrame({

    "Metric": [
        "Test R2",
        "Test RMSE ($)",
        "Test MAE ($)",
        "Test MAPE"
    ],

    "Value": [
        test_r2,
        test_rmse,
        test_mae,
        test_mape
    ]

})

final_report

os.makedirs(
    "reports",
    exist_ok=True
)


final_report.to_csv(
    "reports/final_metrics.csv",
    index=False
)
```
